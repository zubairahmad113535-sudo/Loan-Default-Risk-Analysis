Create AGE group


```python
df["AgeGroup"]=pd.cut(df["Age"],bins=[0, 25 ,35 ,45 ,55 ,100],
labels=["18-25","26-35","36-45","46-55","+56"])

```

```python
age_default=(df.groupby("AgeGroup",observed=True)["Default"].mean()*100).sort_values(ascending=False)
```

```python
print(age_default)
```

```python
df["Income"].describe()
```

Create group of income

```python
df['IncomeGroup'] = pd.qcut(
    df['Income'],
    q=4,
    labels=['Low Income', 'Lower-Middle Income', 'Upper-Middle Income', 'High Income']
)

income_default = (
    df.groupby('IncomeGroup', observed=True)['Default']
      .mean() * 100
).sort_values(ascending=False)

print(income_default)
```

Analyze  default by loan amount

```python
df.groupby("Default")["loan_amount"].mean()
```

Loan to income ratio

```python
df["loan_income"]= df["loan_amount"] / df["Income"]
```

```python
df.groupby("Default")["loan_income"].mean()
```

```python
df["loan_income"].describe()
```

create group of loan income

```python
df['LoanIncomeGroup'] = pd.qcut(
    df['loan_income'],
    q=4,
    labels=['Low', 'Moderate', 'High', 'Very High']
)

loan_income_default = (
    df.groupby('LoanIncomeGroup', observed=True)['Default']
      .mean() * 100
).sort_values(ascending=False)

print(loan_income_default)
```

create group by credit score

```python
df['CreditScoreGroup'] = pd.qcut(
    df['credit_score'],
    q=4,
    labels=['Low', 'Moderate', 'High', 'Very High']
)

credit_default = (
    df.groupby('CreditScoreGroup', observed=True)['Default']
      .mean() * 100
).sort_values(ascending=False)

print(credit_default)
```

```python
df['ReviewScore'] = (
    (df['AgeGroup'] == '18-25').astype(int)
    + (df['IncomeGroup'] == 'Low Income').astype(int)
    + (df['LoanIncomeGroup'] == 'Very High').astype(int)
    + (df['CreditScoreGroup'] == 'Low').astype(int)
    + (df['employment_type'] == 'Unemployed').astype(int)
)

df['ReviewScore'].value_counts().sort_index()
```

```python
df.groupby("ReviewScore")["Default"].mean().mul(100).round(2)
```

```python
df['ReviewCategory'] = pd.cut(
    df['ReviewScore'],
    bins=[-1, 1, 3, 5],
    labels=['Standard', 'Manual Review', 'High Priority Review']
)

df['ReviewCategory'].value_counts().sort_index()
```
